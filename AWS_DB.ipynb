```python
%%sql
CREATE TABLE customers (
    customer_id INT PRIMARY KEY AUTO_INCREMENT,
    customer_name VARCHAR(100) NOT NULL,
    email VARCHAR(100) UNIQUE,
    phone VARCHAR(15) UNIQUE,
    city VARCHAR(50),
    signup_date DATE
);
```

```python
%%sql
CREATE TABLE restaurants (
    restaurant_id INT PRIMARY KEY AUTO_INCREMENT,
    restaurant_name VARCHAR(100) NOT NULL,
    city VARCHAR(50),
    rating DECIMAL(2,1)
);
```

```python
%%sql
CREATE TABLE delivery_partners (
    partner_id INT PRIMARY KEY AUTO_INCREMENT,
    partner_name VARCHAR(100) NOT NULL,
    vehicle_type VARCHAR(30),
    city VARCHAR(50),
    rating DECIMAL(2,1)
);
```

```python
%%sql
CREATE TABLE orders (
    order_id INT PRIMARY KEY AUTO_INCREMENT,

    customer_id INT NOT NULL,
    restaurant_id INT NOT NULL,
    partner_id INT NOT NULL,

    order_date DATE,
    order_status VARCHAR(30),
    payment_method VARCHAR(30),
    total_amount DECIMAL(10,2),
    delivery_time INT,

    FOREIGN KEY (customer_id)
    REFERENCES customers(customer_id)
    ON DELETE CASCADE,

    FOREIGN KEY (restaurant_id)
    REFERENCES restaurants(restaurant_id)
    ON DELETE CASCADE,

    FOREIGN KEY (partner_id)
    REFERENCES delivery_partners(partner_id)
    ON DELETE CASCADE
);
```

```python
%%sql
CREATE TABLE order_items (
    item_id INT PRIMARY KEY AUTO_INCREMENT,

    order_id INT NOT NULL,

    item_name VARCHAR(100),
    quantity INT,
    price DECIMAL(10,2),

    FOREIGN KEY (order_id)
    REFERENCES orders(order_id)
    ON DELETE CASCADE
);
```

```python
%%sql
CREATE TABLE payments (
    payment_id INT PRIMARY KEY AUTO_INCREMENT,

    order_id INT NOT NULL,

    payment_type VARCHAR(30),
    payment_status VARCHAR(30),
    amount DECIMAL(10,2),
    payment_date DATE,

    FOREIGN KEY (order_id)
    REFERENCES orders(order_id)
    ON DELETE CASCADE
);
```

# Question 2 (Easy)

## Scenario

The operations team wants to identify the **top 5 highest-value orders** placed on the platform.

Write a SQL query to display the **top 5 orders** with the highest `total_amount`.

---

## Requirements

- Display the following columns:
  - `order_id`
  - `customer_id`
  - `total_amount`
- Sort the orders by **total_amount** in descending order.
- Display only the **top 5 records**.

---

## Expected Concepts

- `SELECT`
- `ORDER BY`
- `LIMIT`

---

## Expected Output

| order_id | customer_id | total_amount |
|----------|-------------|--------------|
| ... | ... | ... |
| ... | ... | ... |
| ... | ... | ... |
| ... | ... | ... |
| ... | ... | ... |

---

## Hint

1. Sort the records based on `total_amount`.
2. Use **descending order (`DESC`)**.
3. Return only the first **5** rows.

---

**Difficulty Level:** ⭐ Easy

```python
%%sql

select order_id,customer_id,total_amount
from orders
order by total_amount desc
limit 5;
```
