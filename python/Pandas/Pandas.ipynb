# 1. What is Pandas?

**Pandas** is a Python library used to work with structured and tabular data.

For example:

```text
Customer_ID   Name       City        Amount
101           Karthi     Chennai     500
102           Ravi       Bangalore   700
103           Arun       Coimbatore  900
```

This type of data can be easily handled using Pandas.

Pandas provides tools for:

```text
Reading data
Writing data
Cleaning data
Filtering data
Sorting data
Transforming data
Grouping data
Joining data
Handling missing values
Removing duplicates
Working with dates
Working with strings
Aggregating data
ETL processing
```

---

# 2. Why Pandas is Important for Data Engineering

A Data Engineer frequently receives data in formats such as:

```text
CSV
JSON
Excel
Database tables
API responses
Parquet
```

Pandas can act as a processing layer:

```text
Raw Data
   ↓
Pandas
   ↓
Read
   ↓
Inspect
   ↓
Clean
   ↓
Transform
   ↓
Validate
   ↓
Aggregate
   ↓
Export
```

For example:

```text
raw_orders.csv
       ↓
   pandas.read_csv()
       ↓
   DataFrame
       ↓
   Remove duplicates
       ↓
   Handle missing values
       ↓
   Convert data types
       ↓
   Standardize strings
       ↓
   Calculate revenue
       ↓
   Save clean_orders.csv
```

# 81. `merge()` — Joining DataFrames

Pandas `merge()` is similar to SQL `JOIN`.

Example:

```python
customers = pd.DataFrame({
    "customer_id": [1, 2, 3],
    "name": ["Karthi", "Ravi", "Arun"]
})

orders = pd.DataFrame({
    "customer_id": [1, 2, 3],
    "amount": [500, 700, 900]
})
```
